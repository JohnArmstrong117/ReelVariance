```python
import pandas as pd
import numpy as np
```

## Initial Exploration of the MovieLens Dataset

1. Reading in the data from the csv files and looking at the heads

```python
lens_reviews_df = pd.read_csv('Data/MoviLens_32M_Dataset/ml-32m/ratings.csv')
lens_movies_df = pd.read_csv('Data/MoviLens_32M_Dataset/ml-32m/movies.csv')

lens_reviews_df.head()
```

```python
lens_movies_df.head()
```

2. Adding the Calculated Field Variance to the lens_movies_df

```python
movie_rating_variance = lens_reviews_df.groupby('movieId')['rating'].var()

lens_movies_df['rating_variance'] = lens_movies_df['movieId'].map(movie_rating_variance)

lens_movies_df.head()
```

3. Merging the two data frames on the movieId field

```python
lens_combined_df = pd.merge(lens_reviews_df, lens_movies_df, on='movieId', how='left')

lens_combined_df.head()
```

4. Searching for Null Anomalies

```python
rows_with_nulls = lens_combined_df[lens_combined_df.isna().any(axis=1)]

rows_with_nulls.head()
```

5. What is the Average rating accross all reviews?

```python
average_rating = lens_combined_df['rating'].mean()

print(average_rating)
```

6. What is the Average variance across all movies?

```python
average_variance = lens_movies_df['rating_variance'].mean()
highest_variance_movie = lens_movies_df.loc[lens_movies_df['rating_variance'].idxmax()]
lowest_variance_movie = lens_movies_df.loc[lens_movies_df['rating_variance'].idxmin()]

print(average_variance)
print(highest_variance_movie)
print(lowest_variance_movie)
```
